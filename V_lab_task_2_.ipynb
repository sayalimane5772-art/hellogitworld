```python
import pandas as pd
import numpy as np

# Customer Churn Dataset
url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"

df = pd.read_csv(url)

print("Customer Churn Dataset")
print("----------------------")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
df.info()
```

```python
df.describe(include="all").T
```

```python

print(df["Churn"].value_counts())

print("\nChurn Percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Missing values:")
print(df.isnull().sum())

df = df.dropna()

print("\nFinal dataset shape:", df.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

```python

plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Contract Type vs Customer Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="InternetService", hue="Churn")

plt.title("Internet Service vs Customer Churn")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="PaymentMethod", hue="Churn")

plt.title("Payment Method vs Customer Churn")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="tenure",
    hue="Churn",
    bins=30,
    kde=True,
    element="step"
)

plt.title("Tenure vs Customer Churn")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="Churn", y="MonthlyCharges")

plt.title("Monthly Charges vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()
```

```python
churn_rate = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print(churn_rate.round(2))
```

```python
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

print(payment_churn.round(2))
```

```python
print("===== CUSTOMER CHURN ANALYSIS =====")

print("\nOverall Churn Rate:")
print((df["Churn"].value_counts(normalize=True) * 100).round(2))

print("\nChurn Rate by Internet Service:")
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100
print(internet_churn.round(2))

print("\nAverage Monthly Charges:")
print(df.groupby("Churn")["MonthlyCharges"].mean().round(2))
```

```python
print("===== FINAL CUSTOMER CHURN FINDINGS =====")

print("\n1. Overall Churn Rate:")
print(round((df["Churn"] == "Yes").mean() * 100, 2), "%")

print("\n2. Contract with Highest Churn:")
print("Month-to-month contracts have the highest churn.")

print("\n3. Internet Service:")
print("Fiber optic customers show higher churn than DSL customers.")

print("\n4. Payment Method:")
print("Electronic check customers show relatively higher churn.")

print("\n5. Tenure:")
print("Customers with shorter tenure are more likely to churn.")

print("\n6. Monthly Charges:")
print("Customers with higher monthly charges show higher churn.")

print("\n===== RECOMMENDATIONS =====")
print("1. Provide special offers for month-to-month customers.")
print("2. Encourage customers to choose one-year or two-year contracts.")
print("3. Improve service quality for high-churn internet service users.")
print("4. Provide incentives for safer/automatic payment methods.")
print("5. Give loyalty benefits to new customers to improve retention.")
```
