```python
import pandas as pd
import matplotlib.pyplot as plt

print("Geospatial Data Analysis Project")
print("Project started successfully!")
```

```python
data = {
    'City': ['Pune', 'Mumbai', 'Nashik', 'Nagpur', 'Kolhapur'],
    'Latitude': [18.5204, 19.0760, 19.9975, 21.1458, 16.7050],
    'Longitude': [73.8567, 72.8777, 73.7898, 79.0882, 74.2433],
    'Sales': [50000, 80000, 35000, 45000, 30000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Total Sales:", df['Sales'].sum())

print("Average Sales:", df['Sales'].mean())

print("City with Highest Sales:")

print(df.loc[df['Sales'].idxmax(), 'City'])
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Longitude'],
    df['Latitude'],
    s=df['Sales'] / 100,
    alpha=0.7
)

for i, row in df.iterrows():
    plt.text(
        row['Longitude'],
        row['Latitude'],
        row['City']
    )

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Geospatial Sales Analysis")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(df['City'], df['Sales'])

plt.xlabel("City")
plt.ylabel("Sales")
plt.title("City-wise Sales Analysis")

plt.xticks(rotation=45)
plt.show()
```

```python
highest_sales = df.loc[df['Sales'].idxmax()]

print("City with Highest Sales:", highest_sales['City'])
print("Highest Sales Amount:", highest_sales['Sales'])
```
