```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Student dataset
data = {
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Sayali'],
    'Maths': [80, 70, 90, 85, 75],
    'Science': [75, 85, 88, 90, 80],
    'English': [70, 80, 85, 75, 90]
}

df = pd.DataFrame(data)

print("Student Dataset")
display(df)

# 1. Bar Chart
df.set_index('Name')[['Maths', 'Science', 'English']].plot(kind='bar')
plt.title('Student Marks Comparison')
plt.ylabel('Marks')
plt.show()

# 2. Scatter Plot
plt.scatter(df['Maths'], df['Science'])
plt.title('Maths vs Science Marks')
plt.xlabel('Maths Marks')
plt.ylabel('Science Marks')
plt.show()

# 3. Histogram
plt.hist(df['Maths'], bins=5)
plt.title('Maths Marks Distribution')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()

# 4. Box Plot
sns.boxplot(data=df[['Maths', 'Science', 'English']])
plt.title('Subject Marks Box Plot')
plt.show()

# 5. Correlation Heatmap
correlation = df[['Maths', 'Science', 'English']].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Subject Marks Correlation Heatmap')
plt.show()

# 6. Correlation Matrix
print("Correlation Matrix")
display(correlation)
```
