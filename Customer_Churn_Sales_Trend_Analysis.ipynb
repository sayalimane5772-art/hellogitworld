```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample e-commerce sales data
data = {
    'Order_Date': [
        '2026-01-05', '2026-01-12', '2026-02-08',
        '2026-02-15', '2026-03-10', '2026-03-20',
        '2026-04-05', '2026-04-18', '2026-05-09',
        '2026-05-22'
    ],
    'Customer': [
        'Amit', 'Priya', 'Rahul', 'Sneha', 'Amit',
        'Priya', 'Rahul', 'Sneha', 'Amit', 'Priya'
    ],
    'Product': [
        'Laptop', 'Mobile', 'Headphones', 'Laptop', 'Mobile',
        'Laptop', 'Mobile', 'Headphones', 'Laptop', 'Mobile'
    ],
    'Quantity': [1, 2, 3, 1, 2, 1, 2, 4, 1, 2],
    'Price': [50000, 15000, 2000, 50000, 15000,
              50000, 15000, 2000, 50000, 15000]
}

df = pd.DataFrame(data)

# Calculate revenue
df['Revenue'] = df['Quantity'] * df['Price']

# Extract date features
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Month'] = df['Order_Date'].dt.month_name()
df['Year'] = df['Order_Date'].dt.year

print("E-commerce Sales Dataset")
display(df)

# Monthly Sales Trend
monthly_sales = df.groupby('Month')['Revenue'].sum()

plt.figure(figsize=(10, 5))
monthly_sales.plot(kind='bar')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.show()

# Top Customers
customer_sales = df.groupby('Customer')['Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
customer_sales.plot(kind='bar')
plt.title('Customer Sales Analysis')
plt.xlabel('Customer')
plt.ylabel('Revenue')
plt.show()

# Product Sales
product_sales = df.groupby('Product')['Revenue'].sum()

plt.figure(figsize=(8, 5))
product_sales.plot(kind='bar')
plt.title('Product Revenue Analysis')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.show()

# Correlation Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(
    df[['Quantity', 'Price', 'Revenue']].corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Sales Correlation Heatmap')
plt.show()

# Customer Summary
print("Total Revenue:", df['Revenue'].sum())
print("\nSales by Customer:")
display(customer_sales)

print("\nSales by Product:")
display(product_sales)

print("\nBusiness Recommendations:")
print("1. Focus on products generating higher revenue.")
print("2. Give special offers to regular customers.")
print("3. Track monthly sales to identify growth opportunities.")
print("4. Improve marketing for products with lower sales.")
```
