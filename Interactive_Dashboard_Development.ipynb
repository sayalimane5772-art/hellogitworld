```python
import pandas as pd
import matplotlib.pyplot as plt

print("Interactive Dashboard Development")
print("Project Started Successfully!")
```

```python
data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Keyboard'],
    'Sales': [50000, 80000, 30000, 15000, 10000],
    'Quantity': [10, 20, 8, 15, 25]
}

df = pd.DataFrame(data)

display(df)
```

```python
total_sales = df['Sales'].sum()
total_quantity = df['Quantity'].sum()
average_sales = df['Sales'].mean()

print("Total Sales:", total_sales)
print("Total Quantity Sold:", total_quantity)
print("Average Sales:", average_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df['Product'], df['Sales'])

plt.xlabel("Products")
plt.ylabel("Sales Amount")
plt.title("Product Wise Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    df['Sales'],
    labels=df['Product'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Product Sales Distribution")
plt.show()
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x='Product',
    y='Sales',
    color='Product',
    title='Interactive Product Sales Dashboard',
    hover_data=['Quantity']
)

fig.show()
```

```python
print("DASHBOARD SUMMARY")
print("-----------------")
print("Total Sales:", df['Sales'].sum())
print("Total Products:", len(df))
print("Total Quantity:", df['Quantity'].sum())

best_product = df.loc[df['Sales'].idxmax(), 'Product']
print("Best Selling Product:", best_product)
```
